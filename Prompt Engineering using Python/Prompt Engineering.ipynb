# Role Prompting - Tell the LLM what role it should act as

```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()
gemini_api_key = os.getenv("GEMINI_API_KEY")


client = genai.Client(api_key=gemini_api_key)
MODEL = "gemini-3.5-flash-lite"

question = input("Enter User Question: ")

prompt = f"""
You are an expert Python developer.

Answer the user's question as a professional Python developer.
Give a clear and simple explanation with an example when useful.

User Question:
{question}
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```

# XML PROMPTING - Organize the prompt using XML-style tags

```python
from google import genai

client = genai.Client(api_key=gemini_api_key)
MODEL = "gemini-3.5-flash-lite"

user_query = input("Enter User Question: ")

prompt = f"""
<prompt>
    <role>Technical Assistant</role>
    <task>Provide clear, accurate, and practical technical solutions.</task>
    <user_query>{user_query}</user_query>
    <guidelines>
        <item>Use correct technical terminology.</item>
        <item>Be concise.</item>
        <item>Provide examples or code when relevant.</item>
        <item>Do not assume missing information.</item>
    </guidelines>
</prompt>
"""
response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```

# ReAct PROMPTING 
The prompt encourages the model to reason through a problem step-by-step before giving the final answer

```python
from google import genai

client = genai.Client(api_key=gemini_api_key)
MODEL = "gemini-3.5-flash-lite"


question = input("Enter User Question: ")

prompt = f"""
You are an Technical assistant.

Solve the user's problem using this process:

1. Understand the problem.
2. Identify the important information.
3. Determine the best approach.
4. Check the result.
5. Give the final answer.

Do not provide hidden chain-of-thought.
Instead, provide only a short reasoning summary followed by
the final answer.

User Question:
{question}

Format:

Reasoning Summary:
<short explanation>

Final Answer:
<answer>
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```

# SKELETON OF THOUGHTS

# PROMPT CHAINING - Use the output of one LLM prompt as the input to another prompt

```python
from google import genai

client = genai.Client(api_key=gemini_api_key)
MODEL = "gemini-3.5-flash-lite"

question = input("Enter User Question: ")

# ============================
# CHAIN 1: Generate Explanation
# ============================

prompt_1 = f"""
Explain the following user question.

Give the key concepts and important points.
Do not write the final polished answer yet.

User Question:
{question}
"""

response_1 = client.models.generate_content(
    model=MODEL,
    contents=prompt_1
)



intermediate_result = response_1.text

# ============================
# CHAIN 2: Create Final Answer
# ============================

prompt_2 = f"""
Create a clear and simple final answer using the
information generated by the first prompt.

Original User Question:
{question}

Information from Chain 1:
{intermediate_result}

-Give only the final answer.
-Keep it short and simple.
-Maximum 5 sentences.
"""

response_2 = client.models.generate_content(
    model=MODEL,
    contents=prompt_2
)



final_answer = response_2.text

print("\nResponse:")
print(final_answer)
```

# DYNAMIC TEMPLATE - Dynamically change the prompt based on user input

```python
from google import genai

client = genai.Client(api_key=gemini_api_key)
MODEL = "gemini-3.5-flash-lite"


question = input("Enter User Question: ")

style = input(
    "Enter Response Style "
    "(summarize / translate / explain/ code): "
)

templates = {
    "summarize": "Summarize the following text in 3 concise bullet points.",
    "translate": "Translate the following text into French.",
    "explain": "Explain the following concept clearly.",
    "code": "Write only a short Python code for the following task. Do not explain the code"
}

selected_template = templates.get(
    style.lower(),
    templates["explain"]
)

prompt = f"""
You are a helpful AI assistant.

{selected_template}

User Question:
{question}
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```
